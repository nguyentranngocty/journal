## list

```python
ls = [1, 4.5, 'Python', True]
ls
```

```python
ls.append('orange') #last and can duplicate
ls
```

```python
ls.insert(1, 'apple') #insert at index 1
ls
```

```python
ls.pop() #last
ls
```

```python
ls.pop(1)
ls
```

```python
ls.remove('1') #remove by value if value not exist --> error
ls
```

```python
len(ls)
```

```python
# list comprehension
# [f(x) for x in iterable]

sqr_list = [x**2 for x in range(10)]
sqr_list
```

```python
# slicing
a = ls[1:4] # from index 1 to 3
b = ls[1:4:2] # from index 1 to 3 with step 2
c = ls[::] # copy the entire list
d = ls[::-1] # reverse the list
c[0:1] = ['a', 'b', 'c'] # replace the first element with a, b, c
print(a, b, c, d)
```

```python
# enumerate --> [index, value]

for i, v in enumerate(ls):
    print(i, v)
```

```python
# duplicate *
print(ls*2)
print("-"*80)

# concat +
print(ls + ['banana', 'kiwi'])
```

## tuple

```python
t1 = (1, 4.5, 'Python', True)
t2 = (1, )
t3 = 1, 2.3, "Banana", False
t4 = tuple([6, "Banana", False])
t5 = ("Banana", [1,2,3], (1,2,3))
t6 = tuple("convert str to tuple")
```

```python
# can change value of list or set inside tuple
t5[1].append(4)  # This is allowed
t5
```

```python
# update tuple: tuple -> list -> update -> tuple
t1 = list(t1)
t1[0] = 100
t1 = tuple(t1)
t1
```

```python
# for each
for item in t1:
    print(item)
print("-"*80)

# unpack tuple
a, b, c, d = t1
print(a, b, c, d)
print("-"*80)

# index
for i in range(len(t1)):
    print(t1[i])
```

## set

```python
# no index, unique, unordered, mutable
se1 = {1,2, 10, 1, 8, 6}
print(se1)
```

```python
se2 = {1, 4.5, 3, 6, 'Python', True}
se2.add('orange') # last cannot duplicate
se2
```

```python
se3 = set(se2)
se3.remove('orange') # remove by value if value not exist --> error
se3
```

```python
print('Union:', se2.union(se3)) #|
print('Intersection:', se2.intersection(se3)) #&
print('Difference:', se2.difference(se3))
print(se2.issuperset(se3)) # issubset, ...
print(len(se2))
```

```python
clean_dup_list = ['url1', 'url2', 'url1', 'url3', 'url2']

remove_dup_list = list(set(clean_dup_list))
print(remove_dup_list)


set_comprehension = {url for url in clean_dup_list}
print(list(set_comprehension))
```

```python
# sort set: set -> list -> sort
sorted_set = sorted(list(set(clean_dup_list)))
print(sorted_set)
```

## dictionary

```python
dct = {
    'id': '001', 
    'name': 'Ngoc Ty',
    'age': 20,
    'role': 'Dev'
}
print(dct)
```

```python
key_ls = [1, 2, 3, 4]
value_ls = ['v1', 'v2', 'v3', 'v4']

# merge dict
dict_merge = dict(zip(key_ls, value_ls))
print(dict_merge)

dict_compre = {k: v for k, v in zip(key_ls, value_ls)}
print(dict_compre)
```

```python
dct['age'] = 19 # modify value
dct['city'] = 'HCM' # add new key-value
dct.pop('role') #(key)
print(dct, len(dct))
```

```python
for k, v in dct.items():
    print(k, v)
print("-"*80)

for i in dct:
    print(i, dct[i])
print("-"*80)

print(dct.keys())
print(dct.values())
```

```python
# check key existence not value
print('role' in dct) 


# get value by key tranh bi loi khong ton tai
print(dct.get('name', 'na')) #(value, alternative_val)
```
