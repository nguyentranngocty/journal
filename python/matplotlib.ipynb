## import lib

```python
import matplotlib.pyplot as plt
from matplotlib import markers # diểm đánh dấu

import numpy as np
```

## cơ bản

- `plt.figure()`: tạo figure 
- `plt.plot()`: vẽ đồ thị (x, y) 
- `plt.xlabel()`
- `plt.ylabel()`
- `plt.title()`
- `plt.grid()`: đường lưới 
- `plt.legend()`: chú thích cho đồ thị 
- `plt.savefig()`: lưu hình
- `plt.show()`


```python
x = np.arange(-5, 5, 0.5)
x
```

```python
y = x**2
y
```

```python
plt.title('y=x^2')
plt.xlabel('x')
plt.ylabel('y')

plt.grid()

plt.plot(
    x, y,
    label='y=x^2',
    marker='o',
    color='orange',
    markersize=0.5,
    mfc='r', # mau diem
    linestyle='-'
)

plt.legend()
plt.show()
```

## nhiều biểu đồ cùng 1 hình

- `.subplot()` có 3 thông số:
    - `nrows`: số lượng hàng trên Figure
    
    - `ncols`: số lượng cột trên Figure
      
    - `plot_number`: vị trí của plot trong Figure

```python
y2 = -x**2

plt.subplot(1, 2, 1)
plt.plot(x, y, 'red')

plt.subplot(1, 2, 2)
plt.plot(x, y2, 'blue')

plt.show()
```

```python
plt.plot(x, y, 'red', label='y=x^2')
plt.plot(x, y2, 'blue', label='y=-x^2')

plt.title('Quadratic Functions')
plt.grid()
plt.legend()

plt.show()
```

## hướng đối tượng

**figure** có thể chứa một hoặc nhiều **axes**. 

**figure** chỉ là khung chứa, **axes** là nơi các hình được vẽ lên. (**không có** ~~fig.plot(), fig.label(),...~~)

```python
fig = plt.figure()
plt.show()
```

```python
# ax = fig.add_axes([left, bottom, width, height])
fig = plt.figure()
fig.add_subplot(111)

ax = fig.add_axes([0, 0.5, 0.5, 0.5], facecolor='blue') #type: ignore
ax.plot(x, y, 'red')
ax.set_xlabel('Trục x')
ax.set_ylabel('Trục y')
ax.set_title('Hình thứ 1')

ax2 = fig.add_axes([0.5, 0, 0.5, 0.5])# type: ignore
ax2.plot(x, y2, 'red')
ax2.set_xlabel('Trục x')
ax2.set_ylabel('Trục y')
ax2.set_title('Hình thứ 2')

plt.grid()
plt.legend()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,5))

x = np.linspace(0, 2, 100)
ax.plot(x, x, label='linear')
ax.plot(x, x**2, label='quadratic')
ax.plot(x, x**3, label='cubic')

ax.set_xlabel('Trục x')
ax.set_ylabel('Trục y')
ax.legend()

plt.show()
```

```python
x = np.arange(-10., 10., 0.2)
y = np.sin(x)

plt.plot(x, y)
plt.show()
```

```python
# Chỉnh lại trục toạ độ 
fig, ax = plt.subplots(figsize=(10,6))

# Biến đường biên bên trên và bên phải -> vô hình 
ax.spines['top'].set_color('none')
ax.spines['right'].set_color('none')

# Di chuyển đường bên dưới vào giữa  & ở vị trí y = 0
ax.xaxis.set_ticks_position('bottom')
ax.spines['bottom'].set_position(('data',0))

# Di chuyển đường bên trái vào giữa & ở vị trí x = 0
ax.yaxis.set_ticks_position('left')
# ax.spines['left'].set_position(('data', 0))

# Hoặc thay ('data', 0) thành 'center'
ax.spines['left'].set_position('center')

ax.plot(x, y)
plt.show()
```

```python
markers_on = [0, 60, 80, 5]
plt.plot(x, y, '--bo', markevery=markers_on)
plt.show()
```

## sample charts

**Line**

```python
x = np.linspace(-np.pi, np.pi, 30)
y = np.sin(x)

plt.plot(x, y, 'red')
plt.show()
```

**Scatter**

```python
x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])
sizes = np.array([20,50,100,200,500,1000,60,90,10,300,600,800,75])

#kích thước s=size // s=10....
plt.scatter(
    x, y, 
    s=sizes, 
    alpha=0.3, # transparency
    color='red'
)
plt.show()
```

**Bar**

```python
x = np.array(["A", "B", "C", "D"])
y = np.array([3, 8, 1, 10])

plt.subplot(131)
plt.bar(x,y,color='r',width=0.5)

plt.subplot(132)
plt.barh(x,y,color='r',height=0.5)

plt.subplot(133)
plt.hist(x, bins=8, color='green', alpha=0.5)

plt.show()
```

**Pie**

```python
plt.pie(y,labels=x, startangle=90) # type: ignore
plt.show()
```
