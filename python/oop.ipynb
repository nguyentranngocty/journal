## arguments

```python
import math

# __: private
# _ : protected for inheritance

class PhanSo:
    def __init__(
        self, # tự hiểu là chính object được khởi tạo --> gọi 
        __tu_so, # private attribute
        __mau_so
    ):
        self.__tu_so = __tu_so
        self.__mau_so = __mau_so

    # getter, setter    
    def get_ps(self):
        return self.__tu_so, self.__mau_so
    def set_ps(self, tu:int, mau:int):
        self.__tu_so = tu
        self.__mau_so = mau
        
    # methods
    def rut_gon(self):
        ucln = math.gcd(self.__tu_so, self.__mau_so)
        self.__tu_so //= ucln
        self.__mau_so //= ucln
        
    def hien_thi(self):
        print(f'{int(self.__tu_so)}/{int(self.__mau_so)}')

    # overloading
    def __str__(self):
        return f"{self.__tu_so}/{self.__mau_so}"
```

```python
ps1 = PhanSo(10, 20)
ps1.hien_thi()
ps1.rut_gon()
print(ps1) #__str__
```

## no arguments

```python
class Person:
    def __init__(self):
        self._info ={
            'name': "Ty",
            'id': '23521758',
            'age': 20
        }
    # overloading
    def __getitem__(self, key):
        return self._info.get(key, None)
    
    def show_info(self):
        for key, value in self._info.items():
            print(f'{key}: {value}')
```

```python
p1 = Person()
print(p1['name']) # overloading
p1.show_info()
```

## inheritance

```python
class Student(Person):
    def __init__(self):
        super().__init__()
        self._info['uni'] = 'UIT'
        self._info['role'] = 'student'
    
    def show_info(self):
        super().show_info()
        print("Studying!")
```

```python
st = Student()
st.show_info()
```
