## import lib

```python
import plotly.express as px
import plotly.graph_objects as go

import plotly.io as pio
#pio.renderers.default = "svg" # "png", "svg", "browser"
pio.renderers.default = 'vscode'
```

```python
print(pio.renderers)
```

## load data

```python
print(px.data.__all__)
```

```python
px.data.gapminder().head(3)
```

```python
df_us = px.data.gapminder().query("country == 'United States'")
df_europe = px.data.gapminder().query("continent == 'Europe' and year == 2007 and pop > 2.e6")
df_samer = px.data.gapminder().query("year == 2007").query("continent == 'Asia'")
```

```python
df_tips = px.data.tips()
df_iris = px.data.iris()
df_stocks = px.data.stocks()
```

## sample charts

**Line**

```python
px.line(
    df_stocks, 
    x='date', 
    y='GOOG', 
    labels={'x':'Date', 'y':'Price'}
)
```

```python
px.line(
    df_stocks, 
    x='date', 
    y=['GOOG','AAPL'], 
    labels={'x':'Date', 'y':'Price'},
    title='Apple Vs. Google'
)
```

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=df_stocks.date, 
        y=df_stocks.AAPL, 
        mode='lines', 
        name='Apple'
    )
)
fig.add_trace(
    go.Scatter(
        x=df_stocks.date, 
        y=df_stocks.AMZN, 
        mode='lines+markers', 
        name='Amazon'
    )
)
# Dashes : dash, dot, dashdot
fig.add_trace(
    go.Scatter(
        x=df_stocks.date, 
        y=df_stocks.GOOG, 
        mode='lines+markers', 
        name='Google',
        line=dict(color='firebrick', width=2, dash='dashdot')
    )
)

fig.update_layout(
    # Shows gray line without grid, styling fonts, linewidths and more
    xaxis=dict(
        showline=False,
        showgrid=True,
        showticklabels=True,
        linecolor='rgb(204, 204, 204)',
        linewidth=2,
        ticks='outside',
        tickfont=dict(
            family='Arial',
            size=12,
            color='rgb(82, 82, 82)',
        ),
    ),
    # Turn off everything on y axis
    yaxis=dict(
        showgrid=True,
        zeroline=True,
        showline=True,
        showticklabels=True,
    ),
    autosize=True,
    margin=dict(
        autoexpand=False,
        l=100,
        r=20,
        t=110,
    ),
    showlegend=True,
    plot_bgcolor='white'
)
```

**Bar**

```python
px.bar(df_us, x='year', y='pop')
```

```python
px.histogram(df_tips, x="total_bill", color="sex")
```

```python
px.bar(
    df_tips, 
    x='day', 
    y='tip', 
    color='sex', 
    title='Tips by Sex on Each Day',
    labels={'tip': 'Tip Amount', 'day': 'Day of the Week'}
)
```

```python
fig = px.bar(df_europe, y='pop', x='country', text='pop', color='country')

fig.update_traces(
    #texttemplate="%{text:.2f}", khó nhìn
    texttemplate="%{text:.2s}",
    textposition="outside"
)

# uniformtext_mode='hide':hide the text if it won't fit
fig.update_layout(uniformtext_minsize=8)

# Rotate labels 45 degrees
fig.update_layout(xaxis_tickangle=-45)
```

**Pie**

```python
px.pie(
    df_samer, 
    values='pop', 
    names='country', 
    title='Population of Asian continent', 
    #color_discrete_sequence=px.colors.sequential.RdBu
)
```

```python
# Customize pie chart
colors = ['blue', 'green', 'black', 'purple', 'red', 'brown']
fig = go.Figure(
    data=[go.Pie(
        labels=['Water','Grass','Normal','Psychic', 'Fire', 'Ground'], 
        values=[110,90,80,80,70,60])
        ]
    )
# Define hover info, text size, pull amount for each pie slice, and stroke
fig.update_traces(
    hoverinfo='label+percent', 
    textfont_size=20,
    textinfo='label+percent',
    pull=[0.1, 0, 0.2, 0, 0, 0],
    marker=dict(colors=colors, line=dict(color='#FFFFFF', width=2)))
```

**Boxplot**

```python
px.box(df_tips, x='sex', y='tip', points='all')
```

```python
px.box(df_tips, x='day', y='tip', color='sex')
```

```python
fig = go.Figure()
fig.add_trace(
    go.Box(
        x=df_tips.sex, 
        y=df_tips.tip, 
        marker_color='blue',
        boxmean='sd'
    )
)
fig.update_layout(
    title="Boxplot", 
    yaxis=dict(gridcolor='rgb(255, 255, 255)',gridwidth=3),
    paper_bgcolor='rgb(243, 243, 243)',
    plot_bgcolor='rgb(243, 243, 243)'
)
```

**Violin Plot**

```python
px.violin(df_tips, y="total_bill", box=True, points='all')
```

```python
px.violin(
    df_tips, 
    y="tip", x="smoker", color="sex", 
    box=True, 
    points="all",
    hover_data=df_tips.columns)
```

```python
fig = go.Figure()
fig.add_trace(go.Violin(x=df_tips['day'][ df_tips['smoker'] == 'Yes' ],
                        y=df_tips['total_bill'][ df_tips['smoker'] == 'Yes' ],
                        legendgroup='Yes', scalegroup='Yes', name='Yes',
                        side='negative',
                        line_color='blue'))
fig.add_trace(go.Violin(x=df_tips['day'][ df_tips['smoker'] == 'No' ],
                        y=df_tips['total_bill'][ df_tips['smoker'] == 'No' ],
                        legendgroup='Yes', scalegroup='Yes', name='No',
                        side='positive',
                        line_color='red'))
```

```python
# Create a heatmap using Seaborn data
import seaborn as sns
flights = sns.load_dataset("flights")
flights

# You can set bins with nbinsx and nbinsy
fig = px.density_heatmap(flights, x='year', y='month', z='passengers', 
                         color_continuous_scale="Viridis")
fig
```

```python
# You can add histograms
fig = px.density_heatmap(flights, x='year', y='month', z='passengers', 
                         marginal_x="histogram", marginal_y="histogram")
fig
```

```python
# Create a 3D scatter plot using flight data
fig = px.scatter_3d(flights, x='year', y='month', z='passengers', color='year',
                   opacity=0.7, width=800, height=400)
fig
```
