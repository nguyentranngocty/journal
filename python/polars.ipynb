**LazyFrame is much more efficient**

```python
data = pl.read_parquet('./data/items.parquet')
data.head(2)
```

```python
from memory_profiler import memory_usage
import time

def eager_execution(df):
    df_filter = df.filter(pl.col("brand") == "WinWinToys")
    df_agg = df_filter.group_by("category_l1").agg(pl.col("price").mean())
    return df_agg

def lazy_frame(df):
    q = (
        df.lazy()
        .filter(pl.col("brand") == "WinWinToys")
        .group_by("category_l1")
        .agg(pl.col("price").mean())
    )
    return q.collect()


def show_result(func):

    start = time.perf_counter()

    mem, result = memory_usage(
        (func,),
        retval=True,
        max_usage=True
    )

    elapsed = time.perf_counter() - start

    print(result)
    print(f"Peak Memory: {mem:.2f} MiB")
    print(f"Elapsed: {elapsed:.4f}s")
    
    
print("Eager Execution")
show_result(lambda: eager_execution(data))
print()

print("Lazy Frame")
show_result(lambda: lazy_frame(data))
```
