## import lib

```python
import seaborn as sns
import scipy as sp
```

## load datasets

```python
print(sns.get_dataset_names())
```

```python
tips = sns.load_dataset("tips")
iris = sns.load_dataset("iris")
titanic = sns.load_dataset("titanic")
planets = sns.load_dataset("planets")
```

```python
tips.head(2)
```

## Sample charts

**Scatter**

```python
sns.scatterplot(
    data=tips,
    x='tip',
    y='total_bill',
    hue="day", # tips by day in different colors
    size='size' # dots sizes based on the amount of tips by day
    # palette="YlGnBu"
)
```

**Histogram**

```python
sns.histplot(
    data=tips['tip'],
    kde=True, # enabled the line chart
    bins=30
)
```

**Bar**

```python
sns.barplot(x="sex", y="tip", data=tips, palette="YlGnBu")
```

```python
tips.head(5)
```

```python
sns.barplot(
    data=tips,
    x='sex',
    y='tip',
    hue='day'
)
```

```python
sns.catplot(
    data=tips,
    kind="bar",
    x="sex",
    y="tip",
    col="time",
    hue="day"
)
```

**Boxplot**

```python
sns.boxplot(x="day", y="tip", data=tips, hue="sex", palette="YlGnBu")
```

```python
sns.boxplot(x="day", y="total_bill", data=tips, hue="sex", palette="YlGnBu")
```

```python
# dodge=True it separated the dots of the gender
sns.stripplot(x="day", y="tip", data=tips, hue="sex", dodge=True, palette="YlGnBu")
```

```python
# dodge=False
sns.stripplot(x="day", y="tip", data=tips, hue="sex", palette="YlGnBu")
```

**joinplot**

```python
sns.jointplot(x="tip", y="total_bill", data=tips, kind="reg")
```

```python
sns.jointplot(x="tip", y="total_bill", data=tips, kind="kde")
```

```python
# fill=True, but shade=True is more preferable.
sns.jointplot(x="tip", y="total_bill", data=tips, kind="kde", fill=True)
```

```python
sns.jointplot(x="tip", y="total_bill", data=tips, kind="kde", fill=True, cmap="YlGnBu")
```

**pairplot**

```python
sns.jointplot(x="tip", y="total_bill", data=tips, kind="hex", cmap="YlGnBu")
```

```python
sns.pairplot(titanic.select_dtypes(['number']), hue="pclass")
```

**Heatmap**

```python
numeric_df = titanic.select_dtypes(include=["int64", "float64"])

numeric_df.corr()
```

```python
sns.heatmap(numeric_df.corr(), annot=True, cmap="YlGnBu")
```

```python
sns.heatmap(titanic.corr(), annot=True, cmap="coolwarm")
```

```python
sns.heatmap(titanic.corr(), annot=True, cmap="icefire")
```
