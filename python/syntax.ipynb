### Loop

```python
# For loop

# 1
#range(start, end, step) [start, end-1]
for i in range(1, 10, 2): 
    print(i, end=' ')
print('\n')

print("-"*80)

# 2
for i in range(11): # 2 i khac nhau vi la enclosing scope var
    i += 3
    print(i, end=',')
print('\n')

print("-"*80)


# 3
items = [1, 3, 'apple', (1, 2, 3), [3, 4, 5], {3}, {1:'key'}]
for item in items:
    print(item)
print()

print("-"*80)

# 4 nested loop => matrix
for i in range(4):
    for j in range(4):
        print(j, end=' ')
    print()
```

```python
# break and continue
import random

times = 10
while times > 0:
    times -= 1
    print(f'{10-times}:', end=' ')
    i = random.randint(1, 10)
    if 1 < i < 5: 
        print()
        continue
    elif i == 9:
        print(i) 
        break
    else: print(i)    
    
# infinite
while True:
    check = int(input('number > 0:')) > 0
    if check: 
        print('bye')
        break
```

### Function

```python
# def def_name():
#    ...
#   return (if having value)
       
def show_info_args(id:str, name:str, *args):
    """
    doc
    """
    print(f'ID: {id}, Name: {name}')
    for arg in args:
        print(f'Arg: {arg}')
        
def show_info_kwargs(id:str, name:str, **kwargs):
    """
    calling args w keywords
    """
    print(f'ID: {id}, Name: {name}')
    for key, value in kwargs.items():
        print(f'{key}: {value}')

# positional arguments        
show_info_args('001', 'Ty', 'Dev', 20, 'HN')

print("-"*50)

show_info_args(*['001', 'Ty', 'Dev']) # luôn luôn unpack * và gán first arg

print("-"*50)

# keyword arguments co the f(id="...", name="...",...)
show_info_kwargs('002', 'Ngoc Ty', role='Tester', age=21, city='HCM') 
```

```python
# lambda function
# lambda params: <Code(params)> 


# simple
sort_list = [-10, 5, 3, 8, -2, 7, 4]
sort_list.sort(key=lambda x: abs(x), reverse=True)
print(sort_list)

# DataFrame
import pandas as pd

series1 = pd.Series([5, -3, 2, -8, 7])
series2 = pd.Series([1, 4, -6, 2, -9])

df = pd.DataFrame({
    'A': series1, 
    'B': series2
})

df['sum'] = df.apply(lambda row: row['A'] + row['B'], axis=1)
df
```

```python
# decorator

# original
def greet(name: str):
    print(f"Hello {name}")
    
# in case need to write log when calling function
def log_greet(func):
    def wrapper(name: str):
        print(f"---start---")
        func(name)
        print(f"---end---")
    return wrapper

@log_greet
def greet(name: str):
    print(f"Hello {name}")
    
greet("Ty")
```
