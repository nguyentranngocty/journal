```python
import sys
sys.stdin = open("input.txt", "r")
sys.stdout = open("output.txt", "w")

# a, b = map(int, input().split())
# print("Hello world!")
# print(a, b)	

if __name__ == '__main__':
    n = int(input())
    v = list(map(int, input().split()))
    
    res = []
    flag = v[0]
    res.append(flag)
    for i in range(1, n):
        if v[i] != flag:
            res.append(v[i])
            flag = v[i]
            
    for i in res:
        print(i, end=' ')
```
